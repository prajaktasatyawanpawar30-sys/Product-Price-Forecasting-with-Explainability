# 03 - Exploratory Data Analysis (EDA)
## Product Price Forecasting with Explainability

### Objective:
Perform graphical data exploration on product price distributions, brand pricing comparisons, category spreads, and feature correlations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/processed/cleaned_data.csv')
df.head()
```

```python
# 1. Target Price Distribution
plt.figure(figsize=(9, 4))
sns.histplot(df['Price'], kde=True, color='#2563eb', bins=25)
plt.title('Product Price Distribution (₹)', fontsize=13, fontweight='bold')
plt.xlabel('Price (₹)')
plt.ylabel('Frequency')
plt.show()
```

```python
# 2. Brand vs Average Selling Price
plt.figure(figsize=(10, 5))
brand_avg = df.groupby('Brand')['Price'].mean().sort_values(ascending=False)
brand_avg.plot(kind='bar', color='#10b981')
plt.title('Mean Price by Brand', fontsize=13, fontweight='bold')
plt.ylabel('Mean Selling Price (₹)')
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. Price Variation Across Product Categories
plt.figure(figsize=(10, 5))
sns.boxplot(x='Category', y='Price', data=df, palette='Set2')
plt.title('Price Spread by Category', fontsize=13, fontweight='bold')
plt.xticks(rotation=45)
plt.show()
```

```python
# 4. Numerical Feature Correlation Matrix
plt.figure(figsize=(8, 6))
num_cols = df.select_dtypes(include=[np.number]).columns
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', cbar=True)
plt.title('Feature Correlation Heatmap', fontsize=13, fontweight='bold')
plt.show()
```
