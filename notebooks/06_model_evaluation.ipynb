# 06 - Model Evaluation & Diagnostics
## Product Price Forecasting with Explainability

### Objective:
Inspect performance metrics (MAE, RMSE, R², MAPE) across all tested models, examine residual plots, and verify actual vs. predicted price plots.

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Image, display

metrics_path = '../outputs/model_comparison_metrics.json'
with open(metrics_path, 'r') as f:
    metrics = json.load(f)

metrics_df = pd.DataFrame(metrics).T
metrics_df
```

```python
# R² Score Chart
plt.figure(figsize=(8, 4))
metrics_df['R2_Score'].plot(kind='bar', color='#3b82f6', edgecolor='black')
plt.title('R² Benchmark Comparison Across Algorithms', fontsize=13, fontweight='bold')
plt.ylabel('R² Score')
plt.xticks(rotation=30)
plt.ylim(0.8, 1.0)
plt.show()
```
