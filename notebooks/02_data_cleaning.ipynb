# 02 - Data Cleaning & Preprocessing
## Product Price Forecasting with Explainability

### Objective:
Clean raw product price data, handle missing values, eliminate duplicate rows, validate numeric ranges, and export cleaned dataset.

```python
import os
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/product_prices.csv')
print('Initial Raw Shape:', df.shape)
df.head()
```

```python
# 1. Check missing values
print('Missing values count:')
print(df.isnull().sum())
```

```python
# 2. Remove duplicate records
dupes = df.duplicated().sum()
print(f'Duplicate rows detected: {dupes}')
df = df.drop_duplicates().reset_index(drop=True)
```

```python
# 3. Target Price Validation (Non-zero and positive)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
df = df[df['Price'] > 0]
print('Cleaned dataset shape:', df.shape)
```

```python
# 4. Save to processed directory
os.makedirs('../data/processed', exist_ok=True)
output_path = '../data/processed/cleaned_data.csv'
df.to_csv(output_path, index=False)
print(f'Successfully saved cleaned data to: {output_path}')
```
